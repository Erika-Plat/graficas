```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.DataFrame({ "mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"], "ventas": [1200, 1800, 1500, 2400, 2200, 3000], "clientes": [40, 55, 48, 70, 65, 90] })
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.barplot( data=df, x="mes", y="ventas", ax=axes[0], color="royalblue" )
axes[0].set_title("Ventas por mes")
axes[0].set_ylabel("Ventas ($)")

sns.lineplot( data=df, x="mes", y="clientes", marker="o", linewidth=3, ax=axes[1], color="darkorange" )
axes[1].set_title("Clientes por mes")
axes[1].set_ylabel("Número de clientes")
plt.suptitle("Dashboard de ventas", fontsize=16)
plt.tight_layout()
plt.show()
```
