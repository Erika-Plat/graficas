```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
peliculas = pd.DataFrame({
"pelicula": [ "Interestelar", "Coco", "Matrix", "Titanic", "Toy Story", "Avatar" ],
"genero": [ "Ciencia ficción", "Animación", "Acción", "Drama", "Animación", "Ciencia ficción" ],
"puntuacion": [8.7, 8.4, 8.7, 7.9, 8.3, 7.8],
"vistas_millones": [15, 22, 18, 30, 25, 35] })
peliculas = peliculas.sort_values( "puntuacion", ascending=False )
plt.figure(figsize=(11, 6))
sns.barplot( data=peliculas, x="puntuacion", y="pelicula", hue="genero", palette="viridis" )
plt.title("Películas mejor valoradas", fontsize=18)
plt.xlabel("Puntuación")
plt.ylabel("")
plt.xlim(0, 10)
plt.legend(title="Género")
plt.show()
```
