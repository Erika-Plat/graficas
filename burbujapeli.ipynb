```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
peliculas = pd.DataFrame({
"pelicula": [ "Interestelar", "Coco", "Matrix", "Titanic", "Toy Story", "Avatar" ],
"genero": [ "Ciencia ficción", "Animación", "Acción", "Drama", "Animación", "Ciencia ficción" ],
"puntuacion": [8.7, 8.4, 8.7, 7.9, 8.3, 7.8],
"vistas_millones": [15, 22, 18, 30, 25, 35] })
plt.figure(figsize=(12, 7))
sns.scatterplot( data=peliculas, x="puntuacion", y="vistas_millones", hue="genero", size="vistas_millones", sizes=(150, 1200), alpha=0.8, palette="Set2" )
for _, pelicula in peliculas.iterrows():
    plt.text(
pelicula["puntuacion"] + 0.02,
pelicula["vistas_millones"] + 0.5,
pelicula["pelicula"],
fontsize=10 )
plt.title( "Puntuación y popularidad de películas", fontsize=18 )
plt.xlabel("Puntuación de usuarios")
plt.ylabel("Personas que la vieron — millones")
plt.xlim(7.5, 9.1)
plt.ylim(0, 42)
plt.legend( title="Género / Vistas", bbox_to_anchor=(1.02, 1), loc="upper left" )
plt.tight_layout()
plt.show()
```
